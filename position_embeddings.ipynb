Class POSITIONAL ENCODING

```python
import torch
import torch.nn as nn 

class PositionalEncoding(nn.Module):
    def __init__(self, d_model: int, max_seq_len: int) -> None:
        super().__init__()
        self.max_seq_len = max_seq_len
        self.d_model = d_model

    def forward(self):
        even = torch.arange(0,self.d_model,2).float()
        denominator = torch.pow(10000, even/self.d_model)
        position = torch.arange(self.max_seq_len).reshape(self.max_seq_len, 1)
        even_PE = torch.sin(position/denominator)
        odd_PE = torch.cos(position/denominator)
        stacked = torch.stack([even_PE, odd_PE], dim=2)
        PE = torch.flatten(stacked, start_dim=1, end_dim=2)
        return PE
```

```python
pe = PositionalEncoding(d_model=6, max_seq_len=10)
pe.forward()
```
